# 20. Validate JSON Payloads

**Domain:** Integration Safety  
**Use case:** Protect downstream workflows from incomplete input


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
def require_keys(payload: dict, keys: list[str]) -> dict:
    missing = [key for key in keys if key not in payload]
    if missing:
        raise ValueError(f"Missing: {missing}")
    return payload

require_keys({"case_id": "INC-101", "status": "Open"}, ["case_id", "status"])
```
