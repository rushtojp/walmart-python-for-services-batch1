# 36. Apply Role-Based Authorization

**Domain:** Access Control  
**Use case:** Restrict high-impact agent tools by role


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
PERMISSIONS = {
    "agent": {"read_case", "update_case"},
    "manager": {"read_case", "update_case", "approve"}
}

def allowed(role: str, action: str) -> bool:
    return action in PERMISSIONS.get(role, set())

allowed("manager", "approve")
```

This pattern can be wrapped as a script, notebook function, workflow step, API endpoint, or governed tool used by an enterprise agent.
