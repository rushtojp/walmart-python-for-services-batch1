# 16. Create Age Buckets

**Domain:** Queue Analytics  
**Use case:** Group backlog cases into operational age bands


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
def age_bucket(days: int) -> str:
    if days <= 1: return "0-1 days"
    if days <= 3: return "2-3 days"
    if days <= 7: return "4-7 days"
    return "8+ days"

{days: age_bucket(days) for days in [0, 2, 5, 12]}
```
