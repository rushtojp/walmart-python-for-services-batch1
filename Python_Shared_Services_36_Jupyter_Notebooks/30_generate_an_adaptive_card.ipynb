# 30. Generate an Adaptive Card

**Domain:** Teams Integration  
**Use case:** Create a JSON payload for an approval card


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
def approval_card(title: str, case_id: str) -> dict:
    return {"type": "AdaptiveCard", "version": "1.5",
            "body": [{"type": "TextBlock", "text": title, "wrap": True}],
            "actions": [{"type": "Action.Submit", "title": "Approve",
                         "data": {"case_id": case_id, "decision": "approve"}}]}

approval_card("Approve access request", "REQ-101")
```
