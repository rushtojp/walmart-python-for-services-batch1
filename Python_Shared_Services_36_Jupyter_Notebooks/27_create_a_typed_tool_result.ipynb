# 27. Create a Typed Tool Result

**Domain:** Agent Tools  
**Use case:** Return predictable structured output from an agent tool


## Learning objectives

```python
from dataclasses import dataclass
from typing import Any

@dataclass
class ToolResult:
    ok: bool
    message: str
    data: dict[str, Any] | None = None

ToolResult(True, "Case found", {"case_id": "INC-101"})
```
