# 11. Calculate Ticket Priority

**Domain:** Support Triage  
**Use case:** Apply a consistent impact and urgency matrix

```python
def priority_score(impact: int, urgency: int) -> str:
    score = impact * urgency
    return "P1" if score >= 9 else "P2" if score >= 4 else "P3"

priority_score(3, 3)
```
