# 22. Post a Case Update

**Domain:** API Integration  
**Use case:** Send a controlled update to a service platform


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
# Requires: pip install requests
import requests

def post_update(url: str, token: str, payload: dict) -> dict:
    response = requests.post(url, json=payload,
        headers={"Authorization": f"Bearer {token}"}, timeout=10)
    response.raise_for_status()
    return response.json()

# Example only: post_update("https://service.example/api/cases/101", "TOKEN", {"status": "Resolved"})
```
