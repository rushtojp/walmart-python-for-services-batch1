# 09. Build Onboarding Tasks

**Domain:** Employee Lifecycle  
**Use case:** Generate a role-specific onboarding checklist


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
def onboarding_tasks(role: str) -> list[str]:
    base = ["Create account", "Assign manager", "Share policy pack"]
    extras = {
        "support agent": ["Grant queue access", "Assign knowledge training"],
        "manager": ["Grant approval access"]
    }
    return base + extras.get(role.lower(), [])

onboarding_tasks("support agent")
```
