# 31. Calculate Average Resolution Time

**Domain:** Operational Analytics  
**Use case:** Calculate mean handling time for closed cases


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
from datetime import datetime

def avg_resolution_hours(cases: list[dict]) -> float:
    hours = [(case["closed"] - case["created"]).total_seconds() / 3600
             for case in cases if case.get("closed")]
    return round(sum(hours) / len(hours), 1) if hours else 0.0

cases = [{"created": datetime(2026,8,24,9), "closed": datetime(2026,8,24,12)}]
avg_resolution_hours(cases)
```
