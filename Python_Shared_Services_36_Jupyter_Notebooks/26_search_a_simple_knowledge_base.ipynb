# 26. Search a Simple Knowledge Base

**Domain:** Knowledge Retrieval  
**Use case:** Rank FAQ entries through shared query terms


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
def kb_search(query: str, articles: list[dict], limit: int = 3) -> list[dict]:
    terms = set(query.lower().split())
    scored = [(len(terms & set(article["text"].lower().split())), article)
              for article in articles]
    return [article for score, article in sorted(scored, key=lambda x: x[0], reverse=True)
            if score][:limit]

articles = [{"title": "Leave", "text": "check leave balance and request leave"},
            {"title": "Password", "text": "reset a forgotten password"}]
kb_search("leave balance", articles)
```
