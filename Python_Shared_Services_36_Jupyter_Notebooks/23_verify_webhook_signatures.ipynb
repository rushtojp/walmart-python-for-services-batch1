# 23. Verify Webhook Signatures

**Domain:** Webhook Security  
**Use case:** Authenticate an incoming workflow event


## Learning objectives

```python
import hmac
import hashlib

def valid_signature(body: bytes, secret: bytes, supplied: str) -> bool:
    digest = hmac.new(secret, body, hashlib.sha256).hexdigest()
    return hmac.compare_digest(digest, supplied)

body, secret = b'{"event":"case.updated"}', b'demo-secret'
signature = hmac.new(secret, body, hashlib.sha256).hexdigest()
valid_signature(body, secret, signature)
```
