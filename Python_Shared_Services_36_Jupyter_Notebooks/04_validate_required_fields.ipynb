# 04. Validate Required Fields

**Domain:** Data Validation  
**Use case:** Identify missing fields in intake forms and case records


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
def missing_fields(row: dict, required: list[str]) -> list[str]:
    return [key for key in required if not str(row.get(key, "")).strip()]

case = {"employee_id": "EMP-123456", "category": "Leave", "description": ""}
missing_fields(case, ["employee_id", "category", "description"])
```
