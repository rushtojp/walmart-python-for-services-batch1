# 28. Apply a Human Approval Gate

**Domain:** Responsible Automation  
**Use case:** Stop sensitive or high-value actions for review


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
def approval_required(action: str, amount: float = 0) -> bool:
    sensitive = {"terminate_user", "release_payment"}
    return action in sensitive or amount >= 10000

approval_required("release_payment", 2500)
```
