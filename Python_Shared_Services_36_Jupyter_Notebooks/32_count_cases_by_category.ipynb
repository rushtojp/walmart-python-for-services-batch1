# 32. Count Cases by Category

**Domain:** Operational Analytics  
**Use case:** Build category totals for a queue summary


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
from collections import Counter

def count_by_category(cases: list[dict]) -> Counter:
    return Counter(case.get("category", "Unknown") for case in cases)

cases = [{"category": "HR"}, {"category": "IT"}, {"category": "HR"}, {}]
count_by_category(cases)
```
