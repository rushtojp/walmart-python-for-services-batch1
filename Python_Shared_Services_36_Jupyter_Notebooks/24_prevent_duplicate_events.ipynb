# 24. Prevent Duplicate Events

**Domain:** Webhook Reliability  
**Use case:** Use an idempotency key to avoid repeated processing


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
processed: set[str] = set()

def should_process(event_id: str) -> bool:
    if event_id in processed:
        return False
    processed.add(event_id)
    return True

[should_process("EVT-1"), should_process("EVT-1")]
```
