# 21. Perform an HTTP GET

**Domain:** API Integration  
**Use case:** Retrieve a service record with authentication and a timeout


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
# Requires: pip install requests
import requests

def get_record(url: str, token: str) -> dict:
    response = requests.get(url,
        headers={"Authorization": f"Bearer {token}"}, timeout=10)
    response.raise_for_status()
    return response.json()

# Example only: get_record("https://service.example/api/cases/101", "TOKEN")
```
