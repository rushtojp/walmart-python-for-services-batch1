# 12. Route Requests by Keyword

**Domain:** Agent Routing  
**Use case:** Route common intents to specialist service agents


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
ROUTES = {"password": "identity", "leave": "hr", "invoice": "finance"}

def route(text: str) -> str:
    normalized = text.lower()
    return next((team for word, team in ROUTES.items() if word in normalized), "general")

route("I need help with my leave balance")
```
