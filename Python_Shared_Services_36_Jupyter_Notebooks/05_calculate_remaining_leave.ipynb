# 05. Calculate Remaining Leave

**Domain:** HR Services  
**Use case:** Calculate a non-negative leave balance


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
def remaining_leave(entitlement: float, taken: float) -> float:
    return max(entitlement - taken, 0)

remaining_leave(24, 9.5)
```
