# 15. Deduplicate Tickets

**Domain:** Queue Quality  
**Use case:** Retain one record for each ticket identifier


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
def dedupe_tickets(tickets: list[dict]) -> list[dict]:
    unique = {}
    for ticket in tickets:
        unique[ticket["id"]] = ticket
    return list(unique.values())

tickets = [{"id": "INC-1", "status": "New"}, {"id": "INC-1", "status": "Assigned"}]
dedupe_tickets(tickets)
```
