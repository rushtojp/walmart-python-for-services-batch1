# 13. Calculate an SLA Deadline

**Domain:** SLA Management  
**Use case:** Calculate a basic response target by priority


## Learning objectives

- Understand the business problem.

```python
from datetime import datetime, timedelta

def sla_deadline(created_at: datetime, priority: str) -> datetime:
    hours = {"P1": 1, "P2": 4, "P3": 24}
    return created_at + timedelta(hours=hours[priority])

sla_deadline(datetime(2026, 8, 24, 9, 0), "P2")
```
