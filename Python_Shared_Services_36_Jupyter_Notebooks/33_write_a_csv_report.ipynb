# 33. Write a CSV Report

**Domain:** Reporting  
**Use case:** Export a controlled operational report


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
import csv
from io import StringIO

def csv_report(rows: list[dict]) -> str:
    if not rows: return ""
    output = StringIO()
    writer = csv.DictWriter(output, fieldnames=list(rows[0].keys()))
    writer.writeheader()
    writer.writerows(rows)
    return output.getvalue()

print(csv_report([{"case_id": "INC-1", "status": "Open"}]))
```
