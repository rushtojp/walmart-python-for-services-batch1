# 19. Retry Temporary Failures

**Domain:** Resilience  
**Use case:** Retry a function after transient timeout errors


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
import time

def retry(function, attempts: int = 3, base_delay: float = 0.01):
    for attempt in range(attempts):
        try:
            return function()
        except TimeoutError:
            if attempt == attempts - 1:
                raise
            time.sleep(base_delay * (2 ** attempt))

state = {"calls": 0}
def unstable_demo():
    state["calls"] += 1
    if state["calls"] < 2: raise TimeoutError("Temporary failure")
    return "Success"

retry(unstable_demo)
```
