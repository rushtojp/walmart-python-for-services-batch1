# 03. Validate Employee IDs

**Domain:** Data Validation  
**Use case:** Reject malformed employee identifiers before processing


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
import re

def valid_employee_id(value: str) -> bool:
    return bool(re.fullmatch(r"EMP-\d{6}", value))

values = ["EMP-123456", "EMP-12", "123456"]
{value: valid_employee_id(value) for value in values}
```
