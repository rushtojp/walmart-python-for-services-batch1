# 17. Redact Identifiers

**Domain:** Privacy  
**Use case:** Remove common email and phone-like values before AI processing


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
import re

def redact(text: str) -> str:
    text = re.sub(r"[\w.-]+@[\w.-]+", "[EMAIL]", text)
    return re.sub(r"\b\d{10,12}\b", "[PHONE]", text)

redact("Contact employee@example.com or 9876543210")
```
