# 34. Prepare Chart Series

**Domain:** Reporting  
**Use case:** Convert category counts into ordered chart inputs


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
def chart_series(counts: dict[str, int]) -> tuple[list[str], list[int]]:
    labels = sorted(counts)
    values = [counts[label] for label in labels]
    return labels, values

chart_series({"HR": 12, "IT": 18, "Finance": 5})
```
