# 14. Flag Overdue Cases

**Domain:** SLA Management  
**Use case:** Identify open cases that have exceeded their deadline


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
from datetime import datetime, timezone

def is_overdue(deadline: datetime, closed: bool = False, now: datetime | None = None) -> bool:
    now = now or datetime.now(timezone.utc)
    return not closed and deadline < now

deadline = datetime(2026, 8, 24, 5, 0, tzinfo=timezone.utc)
is_overdue(deadline, now=datetime(2026, 8, 24, 7, 0, tzinfo=timezone.utc))
```
