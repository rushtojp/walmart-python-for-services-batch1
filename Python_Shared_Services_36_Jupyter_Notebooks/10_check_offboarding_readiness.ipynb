# 10. Check Offboarding Readiness

**Domain:** Employee Lifecycle  
**Use case:** Verify that critical closure controls are complete


## Learning objectives

- Understand the business problem.
- Run and adapt the Python pattern.
- Identify validation, privacy, and approval requirements.


```python
def offboarding_ready(record: dict) -> bool:
    checks = ["access_revoked", "assets_returned", "final_pay_reviewed"]
    return all(record.get(key) is True for key in checks)

record = {"access_revoked": True, "assets_returned": True, "final_pay_reviewed": False}
offboarding_ready(record)
```
